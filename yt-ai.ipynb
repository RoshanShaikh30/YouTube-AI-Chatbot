```python
load_dotenv(override=True)

client = Groq(
    api_key=os.getenv("GROQ_API_KEY")
)

print("✅ Groq client initialized!")
```

```python
def get_video_id(url):
    if "youtu.be/" in url:
        return url.split("youtu.be/")[1].split("?")[0]

    elif "youtube.com/watch?v=" in url:
        return url.split("v=")[1].split("&")[0]

    else:
        return None
```

```python
url = input("Enter YouTube URL: ")

video_id = get_video_id(url)

print("Video ID:", video_id)
```

```python
def get_transcript(video_id):
    transcript = YouTubeTranscriptApi().fetch(video_id)

    text = " ".join([entry.text for entry in transcript])

    return text
```

```python
transcript = get_transcript(video_id)

print(transcript[:1000])
```

```python
def summarize_video(transcript):
    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {
                "role": "system",
                "content": "You are a helpful AI assistant that summarizes YouTube videos."
            },
            {
                "role": "user",
                "content": f"""
Summarize the following YouTube video transcript.

Transcript:
{transcript}

Provide:
1. Main Topic
2. Key Points (bullet points)
3. Final Conclusion

Keep the summary concise and easy to understand.
"""
            }
        ]
    )

    return response.choices[0].message.content
```

```python
summary = summarize_video(transcript)

print(summary)
```

```python
def ask_question(transcript, question):

    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {
                "role": "system",
                "content": """
You are an AI assistant.

Answer ONLY using the given transcript.

If the answer is not present in the transcript, reply:
'I couldn't find that information in the video transcript.'
"""
            },
            {
                "role": "user",
                "content": f"""
Transcript:

{transcript}

Question:
{question}
"""
            }
        ]
    )

    return response.choices[0].message.content
```

```python
question = input("Ask a question about the video: ")

answer = ask_question(transcript, question)

print("\nAnswer:\n")
print(answer)
```

```python
while True:

    question = input("\nAsk a question (type 'exit' to quit): ")

    if question.lower() == "exit":
        print("Goodbye!")
        break

    answer = ask_question(transcript, question)

    print("\nAI:")
    print(answer)
```
